# **Topic Introduction**

## ***Time Series Data***

A time series is a sequence of data points collected or recorded at regular time intervals. It can show trends (long-term increase or decrease), seasonality (regular patterns over time), and noise (random variations). Examples include stock prices, weather data, and electricity usage.


## ***Recurrent Neural Network (RNN)***

RNNs are a type of neural network designed for sequence data. Unlike regular neural networks, RNNs can remember previous inputs using a "hidden state." This helps them learn patterns over time. RNNs are used for tasks like time-series forecasting, speech recognition, and language modeling.

**Key features of RNNs**:

  * **Hidden State**: RNNs maintain a hidden state that is updated at each time step based on the input and the previous hidden state. This allows the network to remember information over time.
  * **Parameter Sharing**: The same weights are used across different time steps, which helps in learning patterns in sequences without having to learn a new set of parameters for each position in the input sequence.
  * **Backpropagation Through Time (BPTT)**: A variant of the backpropagation algorithm used to train RNNs that accounts for the sequential nature of the data and the dependencies across time steps.



## ***Vanishing Gradient Problem***

When training RNNs using backpropagation through time (BPTT), the gradients (used to update the model's weights) can become very small. This makes it hard for the model to learn long-term patterns. As a result, standard RNNs often forget important information from earlier in the sequence.

## ***Long Short Term Series (LSTM)***

image.png

> Source: [What is LSTM - Long Short Term Memory?](https://www.geeksforgeeks.org/deep-learning-introduction-to-long-short-term-memory/)

Long Short-Term Memory (LSTM) is a special type of Recurrent Neural Network (RNN) designed to better handle sequential data. While RNNs are good at processing sequences, they struggle to remember information from earlier time steps, especially when the gap between relevant information is large.

LSTM solves this issue by introducing a **memory cell** and special structures called **gates** (input, forget, and output gates). These gates help the model decide which information to keep, update, or forget. As a result, LSTM can remember important information over long periods and is much better at learning long-term dependencies than regular RNNs.


## ***LTSM Architecture***

LSTM (Long Short-Term Memory) is a type of RNN that can **remember important information for a long time** and **forget irrelevant data**. It does this using a special structure made of **memory cells** and **gates**.

---

**Main components of LSTM: Cell State (`Cₜ`)**

* Acts like a **long-term memory**.
* Carries important information across time steps.
* Flows through the sequence with minor changes unless updated by the gates.

---

**The 3 Main Gates:**

a. **Forget Gate**

* Decides **what information to forget**.
* Uses a sigmoid function (output between 0 and 1):

  * 0 → forget completely
  * 1 → keep completely

**Formula:**
`fₜ = σ(Wf · [hₜ₋₁, xₜ] + bf)`

b. **Input Gate**

* Decides **what new information to store** in the cell.
* Has two parts:

  * Sigmoid gate → chooses which values to update.
  * Tanh layer → creates new candidate values to add.

**Formula:**
`iₜ = σ(Wi · [hₜ₋₁, xₜ] + bi)`
`Ĉₜ = tanh(Wc · [hₜ₋₁, xₜ] + bc)`
`Cₜ = fₜ ⊙ Cₜ₋₁ + iₜ ⊙ Ĉₜ`

c. **Output Gate**

* Decides **what to output** as the hidden state (`hₜ`).
* Uses cell state (filtered by `tanh`) and a sigmoid layer.

**Formula:**
`oₜ = σ(Wo · [hₜ₋₁, xₜ] + bo)`
`hₜ = oₜ ⊙ tanh(Cₜ)`

---

**How LSTM works**

1. **Forget Gate** removes unneeded information.
2. **Input Gate** adds new important information.
3. **Cell State** gets updated.
4. **Output Gate** decides what information to pass to the next time step.



| Component   | Role                               |
| ----------- | ---------------------------------- |
| Cell State  | Long-term memory                   |
| Forget Gate | Removes irrelevant info            |
| Input Gate  | Adds new useful info               |
| Output Gate | Produces the output & hidden state |

Many variables with missing values in the dataset show high skewness. Features such as PM2.5 (4.87), NO (4.55), NH3 (11.25), SO2 (4.55), Xylene (5.84), and RF (23.16) are strongly right-skewed, indicating that most values are low but a few are extremely high. This distribution suggests the presence of rare but significant events, like pollution spikes or rainfall. Using simple methods like mean or median imputation on such skewed data could distort the actual distribution, underestimating these rare high values.

Interpolation is a better choice because it estimates missing values based on the trend between known points in time. Since the dataset is time series in nature, most variables are expected to change gradually. Interpolation preserves this temporal continuity and smooths the data realistically, making it more suitable for time-dependent models like LSTM. It avoids sudden jumps or artificial flat values that other imputation methods might introduce, and it helps maintain the underlying patterns crucial for accurate forecasting.


```python
def preprocess_datetime(df):
    # convert to date type
    df['DateTime'] = pd.to_datetime(df['From Date'])

    # set DateTime as index
    df.set_index('DateTime', inplace=True)

    # drop the columns since we wont need it anymore
    df.drop(['From Date', 'To Date'], axis=1, inplace=True)

    # fillna using interpolate
    df.interpolate(method='time', inplace=True)
    df.fillna(method='ffill', inplace=True)
    df.fillna(method='bfill', inplace=True)

    # extract from the DateTime
    df['Year'] = df.index.year
    df['Month'] = df.index.month
    df['Weekday'] = df.index.weekday  # 0 = Monday, 6 = Sunday
    df['Hour'] = df.index.hour
    df['Day'] = df.index.day

    return df

df = preprocess_datetime(df)
```

## ***Window Size***

The **Partial Autocorrelation Function (PACF)** is a tool used to measure how much the past values of a time series are related to the current value, but it removes the influence of other time steps in between.

In simple words, it helps us understand the direct relationship between a current value and a specific past value, without being affected by other previous values. For example, when predicting the stock price for today, PACF helps us know if yesterday’s price (or two days ago, or three days ago) is directly related to today’s price, after excluding the influence of other days.

By looking at the PACF, we can decide which past days' prices (lags) are important for making the prediction, so we can use only the most relevant past information to build a better prediction model.

```python
from statsmodels.tsa.stattools import pacf

df_lags = df['AT (degree C)']
pacf_values = pacf(df_lags, nlags=40)
conf_interval = 1.96 / np.sqrt(len(df_lags))
significant_lags = np.where(np.abs(pacf_values) > conf_interval)[0]

print(f"Significant Lags (PACF > 95% CI): {significant_lags}")
```

```python
plt.figure(figsize=(10, 4))
plt.stem(range(len(pacf_values)), pacf_values)
plt.axhline(y=conf_interval, linestyle='--', color='red', label='95% CI')
plt.axhline(y=-conf_interval, linestyle='--', color='red')
plt.title('Partial Autocorrelation (PACF)')
plt.xlabel('Lag')
plt.ylabel('PACF Value')
plt.legend()
plt.tight_layout()
plt.show()
```

From the PACF plot:

* **Lag 1 and 2** show **strong correlation**.
* **Lags 3 to 20** show **small but statistically insignificant** spikes (within the red dashed confidence interval).
* Beyond lag \~25, most values flatten and stay well within the noise threshold.

**Conclusion**: The **most meaningful lags** are the first **1-2**. Statistically, a window size of 2 or 3 would capture most of the autocorrelated structure. But we'll just stick to the instruction of using a fixed window of 5 previous hours to predict the air quality (AT) 1 hour ahead.

```python
def split_and_scale(df, target='AT (degree C)', sequence_length=5, offset=1):
    features = [col for col in df.columns if col not in ['From Date']]
    target_col = df[[target]]

    train_size = int(len(df) * 0.8)
    val_size = int(len(df) * 0.1)

    train_df = df.iloc[:train_size]
    val_df = df.iloc[train_size:train_size + val_size]
    test_df = df.iloc[train_size + val_size:]

    # scaler for ALL features (X)
    feature_scaler = MinMaxScaler()
    feature_scaler.fit(train_df[features]) # fit only on features

    # scaler for TARGET (y)
    target_scaler = MinMaxScaler()
    target_scaler.fit(train_df[[target]]) # fit ONLY on the target column

    # transform features
    train_scaled_features = pd.DataFrame(feature_scaler.transform(train_df[features]), columns=features, index=train_df.index)
    val_scaled_features = pd.DataFrame(feature_scaler.transform(val_df[features]), columns=features, index=val_df.index)
    test_scaled_features = pd.DataFrame(feature_scaler.transform(test_df[features]), columns=features, index=test_df.index)

    # transform target
    train_scaled_target = pd.DataFrame(target_scaler.transform(train_df[[target]]), columns=[target], index=train_df.index)
    val_scaled_target = pd.DataFrame(target_scaler.transform(val_df[[target]]), columns=[target], index=val_df.index)
    test_scaled_target = pd.DataFrame(target_scaler.transform(test_df[[target]]), columns=[target], index=test_df.index)

    def create_sequences(features_data, target_data):
        X, y = [], []
        for i in range(len(features_data) - sequence_length - offset + 1):
            X.append(features_data.iloc[i:i+sequence_length].values)
            y.append(target_data.iloc[i+sequence_length+offset-1][0])
        return np.array(X), np.array(y)

    X_train, y_train = create_sequences(train_scaled_features, train_scaled_target)
    X_val, y_val = create_sequences(val_scaled_features, val_scaled_target)
    X_test, y_test = create_sequences(test_scaled_features, test_scaled_target)

    return X_train, y_train, X_val, y_val, X_test, y_test, target_scaler, df.columns.get_loc(target), df.columns

X_train, y_train, X_val, y_val, X_test, y_test, scaler, at_col_idx, all_columns = split_and_scale(df)
```

**Why do we need scaling?**

We need to scale the data because the LSTM model might get confused when numbers are too big or too different from each other. If some features are in the thousands and others are in the tens, the model might accidentally focus too much on the really big numbers. Scaling just squishes all the numbers into a similar, small range so everything is treated fairly, helping the model learn better and faster.

**Why MinMaxScaler?**

We use something called `MinMaxScaler` for this. Think of it like taking all the  numbers for one type of data (like temperature) and finding the smallest one and the biggest one. Then, `MinMaxScaler` changes all those numbers so the smallest one becomes 0, the biggest one becomes 1, and all the numbers in between just fit neatly between 0 and 1. It's like resizing a picture to fit on a screen; it doesn't change what's in the picture, just its size. This is good for our LSTM because it keeps the relationships between the original temperature values the same, which is important for understanding how temperature changes over time.

```python
def plot_training_metrics(hist):
    history = hist.history

    train_mae = history.get('mae')
    val_mae = history.get('val_mae')

    train_loss = history.get('loss')
    val_loss = history.get('val_loss')

    train_r2 = history.get('r2_score')  # Use actual key names from print
    val_r2 = history.get('val_r2_score')

    train_mape = history.get('mean_absolute_percentage_error')
    val_mape = history.get('val_mean_absolute_percentage_error')

    epochs = range(1, len(train_loss) + 1)

    plt.figure(figsize=(14, 10))
    plt.style.use('fivethirtyeight')

    # MAE
    plt.subplot(2, 2, 1)
    plt.plot(epochs, train_mae, 'r-', label='Train MAE')
    plt.plot(epochs, val_mae, 'g-', label='Val MAE')
    plt.title('Mean Absolute Error')
    plt.xlabel('Epoch')
    plt.ylabel('MAE')
    plt.legend()
    plt.grid(True)

    # MSE (Loss)
    plt.subplot(2, 2, 2)
    plt.plot(epochs, train_loss, 'r-', label='Train Loss')
    plt.plot(epochs, val_loss, 'g-', label='Val Loss')
    plt.title('Mean Squared Error (Loss)')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)

    # R²
    if train_r2 and val_r2:
        plt.subplot(2, 2, 3)
        plt.plot(epochs, train_r2, 'r-', label='Train R²')
        plt.plot(epochs, val_r2, 'g-', label='Val R²')
        plt.title('R² Score')
        plt.xlabel('Epoch')
        plt.ylabel('R²')
        plt.legend()
        plt.grid(True)

    # MAPE
    if train_mape and val_mape:
        plt.subplot(2, 2, 4)
        plt.plot(epochs, train_mape, 'r-', label='Train MAPE')
        plt.plot(epochs, val_mape, 'g-', label='Val MAPE')
        plt.title('Mean Absolute Percentage Error')
        plt.xlabel('Epoch')
        plt.ylabel('MAPE')
        plt.legend()
        plt.grid(True)

    plt.tight_layout()
    plt.show()
```

```python
def evaluate_model(model, X_data, y_true_scaled, target_scaler, label="Set"):
    y_pred_scaled = model.predict(X_data)

    y_true_scaled = y_true_scaled.reshape(-1, 1)
    y_pred_scaled = y_pred_scaled.reshape(-1, 1)

    y_true = target_scaler.inverse_transform(y_true_scaled)
    y_pred = target_scaler.inverse_transform(y_pred_scaled)

    y_true = y_true.flatten()
    y_pred = y_pred.flatten()

    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    r2  = r2_score(y_true, y_pred)
    mape = mean_absolute_percentage_error(y_true, y_pred) * 100 # calculate MAPE and multiply by 100 for percentage

    print(f"{label:<10} | MAE: {mae:>8.4f} | MSE: {mse:>10.4f} | R²: {r2:>7.4f} | MAPE: {mape:>7.2f}%")
    return mae, mse, r2, mape
```

```python
def plot_prediction(model, test_ds, scaler, at_col_idx, all_columns, title="AT Prediction for the Next 1 Hour"):
    y_pred_scaled = model.predict(test_ds)
    y_test_true_scaled = np.concatenate([y for _, y in test_ds], axis=0)

    y_pred_scaled = y_pred_scaled.reshape(-1, 1)
    y_test_true_scaled = y_test_true_scaled.reshape(-1, 1)

    y_pred = scaler.inverse_transform(
        np.repeat(y_pred_scaled, len(all_columns), axis=1)
    )[:, at_col_idx]

    y_true = scaler.inverse_transform(
        np.repeat(y_test_true_scaled, len(all_columns), axis=1)
    )[:, at_col_idx]

    plt.figure(figsize=(12, 6))
    plt.plot(y_true, label='Actual AT')
    plt.plot(y_pred, label='Predicted AT')
    plt.title(title)
    plt.xlabel("Time (Sample Index)")
    plt.ylabel("AT (°C)")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    return y_true, y_pred
```

```python
def inverse_transform_predictions(y_true_scaled, y_pred_scaled, scaler, at_col_idx, n_features):
    true_zeros = np.zeros((len(y_true_scaled), n_features))
    pred_zeros = np.zeros((len(y_pred_scaled), n_features))

    true_zeros[:, at_col_idx] = y_true_scaled.ravel()
    pred_zeros[:, at_col_idx] = y_pred_scaled.ravel()

    y_true = scaler.inverse_transform(true_zeros)[:, at_col_idx].reshape(-1, 1)
    y_pred = scaler.inverse_transform(pred_zeros)[:, at_col_idx].reshape(-1, 1)

    return y_true, y_pred
```

```python
def plot_residuals(y_true_dict, y_pred_dict):
    num_sets = len(y_true_dict)
    plt.figure(figsize=(12, 4 * num_sets))

    for i, label in enumerate(y_true_dict.keys(), start=1):
        residuals = y_true_dict[label].flatten() - y_pred_dict[label].flatten()
        plt.subplot(num_sets, 1, i)
        plt.plot(residuals, marker='o', linestyle='-', color='teal', alpha=0.7)
        plt.axhline(0, color='red', linestyle='--', linewidth=1)
        plt.title(f"Residuals (Actual - Predicted) for {label} Set")
        plt.xlabel("Sample Index")
        plt.ylabel("Residual (°C)")
        plt.grid(True)

    plt.tight_layout()
    plt.show()
```

```python
def show_plot(history_seq, true_future, prediction, at_col_idx, scaler, all_columns, title="Single Step Prediction"):

    sequence_length = history_seq.shape[0]
    time_steps = list(range(-sequence_length, 0))

    history_scaled = history_seq[:, at_col_idx].reshape(-1, 1)

    history_full = np.zeros((sequence_length, len(all_columns)))
    history_full[:, at_col_idx] = history_scaled.flatten()
    history_at = scaler.inverse_transform(history_full)[:, at_col_idx]

    true_full = np.zeros((1, len(all_columns)))
    true_full[:, at_col_idx] = true_future
    true_at = scaler.inverse_transform(true_full)[:, at_col_idx][0]

    pred_full = np.zeros((1, len(all_columns)))
    pred_full[:, at_col_idx] = prediction
    pred_at = scaler.inverse_transform(pred_full)[:, at_col_idx][0]


    plt.figure(figsize=(6, 3.5))
    plt.plot(time_steps, history_at, ".-", label="History", color="gray", linewidth=1, markersize=4)
    plt.plot(0, true_at, "rx", label="True Future", markersize=8)
    plt.plot(0, pred_at, "go", label="Prediction", markersize=8)
    plt.axvline(0, color='black', linestyle='--', linewidth=0.8)
    plt.title(title, fontsize=10)
    plt.xlabel("Time Step", fontsize=9)
    plt.ylabel("AT (°C)", fontsize=9)
    plt.xticks(fontsize=8)
    plt.yticks(fontsize=8)
    plt.legend(fontsize=8)
    plt.grid(True, linestyle=':', linewidth=0.5)
    plt.tight_layout()
    plt.show()
```

## ***Bidirectional***

A **Bidirectional RNN** is a type of recurrent neural network that processes data in **both directions**, forward (from past to present) and backward (from future to past). This means the model can see the full context of a sequence from **both sides**.

In a standard RNN (or LSTM), the model only has access to **past information** when making predictions. For example, to predict today’s air temperature, it would only use data from previous hours.

But in a **Bidirectional RNN**, the model processes the sequence **in reverse as well**. This allows it to learn not only from the past but also from the "future" context within the sequence, like seeing both what came **before** and what comes **after** a specific time step.

This often helps the model understand the data better and can lead to **improved accuracy**, especially when future context is relevant to the current prediction.



```python
modified_model = Sequential([
    Bidirectional(LSTM(64, return_sequences=True), input_shape=(X_train.shape[1], X_train.shape[2])),
    Dropout(0.2),
    Bidirectional(LSTM(32)),
    Dropout(0.2),
    Dense(1)
])
```

**1. Bidirectional LSTM Layer (64 units)**

* **Purpose**: This is the first processing layer of the model. It looks at each 5-hour window from both directions, **past to future** and **future to past**, allowing it to learn more complete patterns from the sequence.
* **What it does**: It uses `64` memory units in each direction, so a total of `128`. `return_sequences=True` means it sends the full sequence of processed data to the next layer instead of just the final output.

---
**2. Dropout Layer (rate = 0.2)**

* **Purpose**: To help the model generalize better and not memorize the training data.
* **What it does**: Randomly turns off 20% of the outputs from the previous LSTM layer during each training step.

---
 **3. Bidirectional LSTM Layer (32 units)**

* **Purpose**: This is the second LSTM layer. It continues learning from the processed sequence, again looking in both directions.
* **What it does**: Uses `32` units per direction (64 total). Since `return_sequences=False` by default, it only passes on the **final summary** of the whole sequence to the next layer.

---

**4. Dropout Layer (rate = 0.2)**

* **Purpose**: Again helps prevent overfitting by turning off neurons randomly.
* **What it does**: This dropout layer is applied to the output of the second LSTM layer.

---

**5. Dense Layer (1 unit)**

* **Purpose**: This is the final step,  the model’s prediction layer.
* **What it does**: It takes all the learned information and outputs **one single number**, which is the predicted **air temperature one hour later**. Since there’s no activation function, the result is a raw continuous number — perfect for regression.

# **Fine Tuned Model**

```python
def build_model(units_1, units_2, dropout_rate, input_shape):
    model = Sequential([
        Bidirectional(LSTM(units_1, return_sequences=True), input_shape=input_shape),
        Dropout(dropout_rate),
        Bidirectional(LSTM(units_2)),
        Dropout(dropout_rate),
        Dense(1)
    ])
    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='mse',
        metrics=[
            'mae',
            tf.keras.metrics.R2Score(name="r2_score"),
            tf.keras.metrics.MeanAbsolutePercentageError(name="mape")
        ]
    )
    return model
```

Now, we're **fine-tuning a Bidirectional LSTM model** by systematically testing different combinations of hyperparameters to find the best architecture for predicting air temperature. Specifically, we are tuning:

* **LSTM units in the first layer (`units_1`)** – how many memory cells the first Bidirectional LSTM layer has.
* **LSTM units in the second layer (`units_2`)** – controls the learning capacity of the second Bidirectional LSTM layer.
* **Dropout rate (`dropout_rate`)** – the fraction of neurons randomly turned off during training to prevent overfitting.


```python
results = []

combinations = [
    (32, 16, 0.2),
    (32, 32, 0.3),
    (64, 16, 0.3),
    (64, 32, 0.2)
]

for units_1, units_2, dropout_rate in combinations:
    print(f"\nTraining with: units_1={units_1}, units_2={units_2}, dropout={dropout_rate}")
    model = build_model(units_1, units_2, dropout_rate, (X_train.shape[1], X_train.shape[2]))

    history = model.fit(
        train_ds,
        validation_data=val_ds,
        epochs=50,
        batch_size=32,
        callbacks=[early_stopping],
        verbose=0
    )

    val_metrics = model.evaluate(val_ds, verbose=0)
    val_loss, val_mae, val_r2, val_mape = val_metrics
    results.append({
        "units_1": units_1,
        "units_2": units_2,
        "dropout": dropout_rate,
        "val_loss": val_loss,
        "val_mae": val_mae,
        "val_r2": val_r2,
        "val_mape": val_mape,
        "history": history,
        "model": model
    })
```

```python
sorted_results = sorted(results, key=lambda x: x['val_mae'])

best_model = sorted_results[0]['model']
best_history = sorted_results[0]['history']

print("Best configuration:")
print(f"units_1: {sorted_results[0]['units_1']}, units_2: {sorted_results[0]['units_2']}, dropout: {sorted_results[0]['dropout']}")
```

Yes, this problem is not only possible but actually already framed as a **multivariate time series** problem. In time series terminology, “multivariate” refers to using more than one feature (variable), all of which are observed over the same time intervals. In this case of predicting the air temperature one hour ahead, we are not only using the temperature (AT), but also other features like relative humidity (RH), pressure, and concentrations of gases like CO, NOx, and NO₂. All of these are recorded hourly and are interrelated.

Using a multivariate approach becomes relevant when the target variable (in this case, temperature) depends on multiple influencing factors. For instance, pressure and humidity can significantly affect future temperature, and this relationship cannot be captured effectively if only a single variable is used. By incorporating all available features, the model can learn complex patterns and interactions between variables, leading to better predictive performance. Multivariate models often outperform univariate ones because they take advantage of the extra information provided by related features.

To determine whether a problem is suitable for a multivariate approach, we should check whether multiple relevant features are consistently recorded over time and whether they are expected to influence the target variable. If so, then a multivariate method is not just possible, it is often **more appropriate and more effective**.
